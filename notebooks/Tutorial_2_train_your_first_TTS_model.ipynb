```python
!pip install edge-tts nest_asyncio

import edge_tts
import asyncio
import nest_asyncio
from google.colab import files
from IPython.display import Audio

nest_asyncio.apply()

async def text_dosyasindan_ses_olustur(text_dosya_yolu, ses_dosya_adi="output.mp3", ses="tr-TR-AhmetNeural"):
    """
    Text dosyasından metin okuyup ses dosyası oluşturur
    """
    # Text dosyasını oku
    with open(text_dosya_yolu, 'r', encoding='utf-8') as f:
        metin = f.read()

    print(f"📄 Metin okundu: {len(metin)} karakter")
    print(f"🎤 Ses oluşturuluyor (İdeal hız: +15%)...")

    # Ses ayarları - ideal hız
    communicate = edge_tts.Communicate(
        metin,
        ses,
        rate="+15%",     # İdeal okuma hızı
        volume="+0%"
    )
    await communicate.save(ses_dosya_adi)

    print(f"✅ Ses dosyası hazır: {ses_dosya_adi}")

    return ses_dosya_adi

# DOSYA YÜKLEME MENÜSÜ
print("📁 Lütfen text dosyanızı seçin (.txt)")
uploaded = files.upload()

# Yüklenen dosyayı al
dosya_adi = list(uploaded.keys())[0]
print(f"\n✅ Dosya yüklendi: {dosya_adi}")

# Sesi oluştur (Ahmet sesi + İdeal hız)
ses_dosyasi = await text_dosyasindan_ses_olustur(dosya_adi, "sesli_metin.mp3", "tr-TR-AhmetNeural")

# Dinle
print("\n🔊 Ses dosyası oynatılıyor...")
display(Audio(ses_dosyasi, autoplay=True))

# İndirme seçeneği
print("\n💾 Ses dosyasını indirmek için aşağıdaki komutu çalıştırın:")
print("files.download('sesli_metin.mp3')")
```
