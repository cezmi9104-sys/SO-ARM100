```python
!pip install git+https://github.com/suno-ai/bark.git

from bark import SAMPLE_RATE, generate_audio, preload_models
from scipy.io.wavfile import write as write_wav
from IPython.display import Audio
from google.colab import files
import numpy as np
import torch # Add torch import for the fix
import functools # Import functools for patching

# Modelleri yükle (ilk seferinde zaman alır)
print("🔄 Modeller yükleniyor...")

# Patch torch.load to force weights_only=False for compatibility with older models
# This bypasses the UnpicklingError when loading models with PyTorch 2.6+
_original_torch_load = torch.load
@functools.wraps(_original_torch_load)
def _patched_torch_load(*args, **kwargs):
    if 'weights_only' not in kwargs:
        kwargs['weights_only'] = False
    return _original_torch_load(*args, **kwargs)
torch.load = _patched_torch_load

# The previous fix (add_safe_globals) is no longer strictly necessary with the patch, but can be kept.
# torch.serialization.add_safe_globals([np.core.multiarray.scalar])

preload_models()

# Dosya yükleme
print("\n📁 Text dosyanızı seçin:")
uploaded = files.upload()
dosya_adi = list(uploaded.keys())[0]

with open(dosya_adi, 'r', encoding='utf-8') as f:
    metin = f.read()

print(f"\n📄 Metin: {len(metin)} karakter")

# TÜRKÇE için özel speaker
# v2/tr_speaker_X formatında Türkçe konuşmacılar var
print("🎤 Ses oluşturuluyor...")

audio_array = generate_audio(
    metin,
    history_prompt="v2/tr_speaker_1"  # Türkçe konuşmacı
)

# Kaydet
write_wav("bark_output.wav", SAMPLE_RATE, audio_array)
print("✅ Ses hazır!")

# Dinle
Audio("bark_output.wav", autoplay=True)
```
